```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("../data/data_application_1.csv")
```

```python
counts = data.groupby("id")["id"].transform("size")
data = data[counts >= 100].copy()
```

```python
MAX_NUM_OBS = 116
NUM_SUBJECTS = data["id"].nunique()
```

```python
emp_data = np.full((NUM_SUBJECTS, MAX_NUM_OBS, 5), np.nan)
```

```python
for i, id, in enumerate(data.id.unique()):
    person_data = data.loc[data.id == id]
    num_obs = len(person_data.rt)
    rt = person_data.rt
    resp = person_data.resp
    stim_type = person_data.stim_type * 1.0 - 0.5 # effect coding
    truth = person_data.factual_truth * 1.0 - 0.5 # effect coding
    interaction = stim_type * truth
    emp_data[i, :num_obs, :] = np.c_[rt, resp, stim_type, truth, interaction]
```

```python
np.save("../data/data_application_1.npy", emp_data)
```
