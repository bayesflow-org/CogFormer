```python
import numpy as np
import pandas as pd
```

## DDM Data

Data is from a statement truthness judgement task (Stump, Schumacher, Voss, Klauer, 2026)

`stim_type`, `truth`, and `interaction` are regressed on drift rate, threshold, bias, ndt and ndt variabilty

```python
ddm_data = pd.read_csv("data_application_ddm.csv")
ddm_data
```

```python
counts = ddm_data.groupby("id")["id"].transform("size")
ddm_data = ddm_data[counts >= 100].copy()
```

```python
max_num_obs = 116
num_subjects = ddm_data["id"].nunique()
```

```python
ddm_emp_data = np.full((num_subjects, max_num_obs, 5), np.nan)
```

```python
for i, id, in enumerate(ddm_data.id.unique()):
    person_data = ddm_data.loc[ddm_data.id == id]
    num_obs = len(person_data.rt)
    rt = person_data.rt
    resp = person_data.resp
    stim_type = person_data.stim_type * 1.0 - 0.5 # effect coding
    truth = person_data.factual_truth * 1.0 - 0.5 # effect coding
    interaction = stim_type * truth
    ddm_emp_data[i, :num_obs, :] = np.c_[rt, resp, stim_type, truth, interaction]
```

```python
np.save("data_application_ddm.npy", ddm_emp_data)
```

## RDM Data

Data is from a three-alternative Stroop task (Enkavi et al., 2018)

drift(accumulator == word) = base_drift + drift_boost_1

drift(accumulator == color) = base_drift + drift_boost_2

```python
rdm_data = pd.read_csv("data_application_rdm.csv")
rdm_data
```

```python
max_num_obs = 192
num_subjects = rdm_data["id"].nunique()
```

```python
rdm_emp_data = np.full((num_subjects, max_num_obs, 4), np.nan)
```

```python
for i, id, in enumerate(rdm_data.id.unique()):
    person_data = rdm_data.loc[rdm_data.id == id]
    num_obs = len(person_data.rt)
    rt = person_data.rt
    resp = person_data.resp
    word = person_data.word
    color = person_data.color
    rdm_emp_data[i, :num_obs, :] = np.c_[rt, resp, word, color]
```

```python
np.save("data_application_rdm.npy", rdm_emp_data)
```

## CDM Data

Data is from a Gabor patch orientation judgment task (Kvam, 2019)

`speed_condition`, `difficulty` and their interaction is regressed on drift length, threshold, and non-decision time

```python
cdm_data = pd.read_csv("../data/data_application_cdm.csv")
cdm_data
```

```python
max_num_obs = 599
num_subjects = cdm_data["id"].nunique()
```

```python
cdm_emp_data = np.full((num_subjects, max_num_obs, 5), np.nan)
```

```python
for i, id, in enumerate(cdm_data.id.unique()):
    person_data = cdm_data.loc[cdm_data.id == id]
    num_obs = len(person_data.rt)
    rt = person_data.rt
    resp = person_data.resp_deviation
    speed_condition = person_data.speed_condition * 1.0 - 0.5 # effect coding
    difficulty = person_data.difficulty
    interaction = difficulty * speed_condition # needs fixing if difficulty not continuous
    cdm_emp_data[i, :num_obs, :] = np.c_[rt, resp, speed_condition, difficulty, interaction]
```

```python
np.save("data_application_cdm.npy", cdm_emp_data)
```
